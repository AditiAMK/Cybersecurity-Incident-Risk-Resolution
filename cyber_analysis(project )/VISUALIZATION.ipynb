```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/Global_Cybersecurity_Threats_2015-2024 2.csv')

# Basic information
print("Rows and columns:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Remove extra spaces from text columns
text_columns = df.select_dtypes(include='object').columns

for column in text_columns:
    df[column] = df[column].str.strip()

# Convert numeric columns to numeric format
numeric_columns = [
    'Year',
    'Financial Loss (in Million $)',
    'Number of Affected Users',
    'Incident Resolution Time (in Hours)'
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors='coerce')

# Check final dataset
print("\nCleaned dataset shape:", df.shape)

print("\nFinal missing values:")
print(df.isnull().sum())

# Save cleaned data
df.to_csv('/content/cybersecurity_threats_cleaned.csv', index=False)

print("\nCleaned file created successfully.")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['Financial Loss (in Million $)'], bins=20)

plt.title('Distribution of Financial Loss')
plt.xlabel('Financial Loss (Million $)')
plt.ylabel('Number of Incidents')

plt.savefig('/content/financial_loss_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
attack_loss = df.groupby('Attack Type')['Financial Loss (in Million $)'].mean().sort_values()

plt.figure(figsize=(8,5))
attack_loss.plot(kind='bar')

plt.title('Average Financial Loss by Attack Type')
plt.xlabel('Attack Type')
plt.ylabel('Average Financial Loss (Million $)')
plt.xticks(rotation=45)

plt.show()
```
